```python
import pandas as pd
df = pd.read_csv("../data/raw/naukri_com-job_sample.csv")
print(df.shape)
df.head(5)
```

```python
(df.isna().mean() * 100).round(1).sort_values(ascending=False)
```

```python
df_clean = df.drop(columns=["site_name", "numberofpositions"])
print(df_clean.shape)
```

Dropped `site_name` (81.9% missing) and `numberofpositions` (79.7% missing): too empty to fill reliably.

```python
print("duplicate jobid:", df_clean["jobid"].duplicated().sum())
print("duplicate uniq_id:", df_clean["uniq_id"].duplicated().sum())
print("fully identical rows:", df_clean.duplicated().sum())
print("same content, different ids:", df_clean.drop(columns=["jobid", "uniq_id"]).duplicated().sum())
```

```python
dups = df_clean[df_clean["jobid"].duplicated(keep=False)].sort_values("jobid")
dups[["jobid", "company", "jobtitle", "joblocation_address"]].head(10)
```

```python
content_cols = [c for c in df_clean.columns if c not in ["jobid", "uniq_id"]]
df_clean = df_clean.drop_duplicates(subset=content_cols)
print(df_clean.shape)
```
